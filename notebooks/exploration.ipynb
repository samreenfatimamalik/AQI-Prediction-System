```python
# bcz traffic can be varry from day to day(eg less on weekends)
df2["day_of_week"] = df2["time"].dt.dayofweek
```

```python
# month column is added bcz in pakistan,,,pollution vary from months to months 
df2["month"] = df2["time"].dt.month
```

```python
df2.head()
```

### Implementing Lag Features

```python
df2['PM2.5_lag_1'] = df2['PM2.5'].shift(1)
df2['PM2.5_lag_24'] = df2['PM2.5'].shift(24)
```

```python
df2[['PM2.5','PM2.5_lag_1','PM2.5_lag_24']].head(30)
```

### Rolling Average

```python
df2['PM2.5_rolling_3'] = (
    df2['PM2.5']
    .rolling(window=3)
    .mean()
)

df2['PM2.5_rolling_24'] = (
    df2['PM2.5']
    .rolling(window=24)
    .mean()
)
```

```python
df3 = pd.read_csv('../data/processed_aqi.csv')

# it is making sure time is sorted (very important for time series)
df3['time'] = pd.to_datetime(df3['time'])
df = df3.sort_values('time').reset_index(drop=True)

# Create target columns: PM2.5 value N hours in the future
df3['target_24h'] = df3['PM2.5'].shift(-24)   # PM2.5, 24 hours ahead
df3['target_48h'] = df3['PM2.5'].shift(-48)   # PM2.5, 48 hours ahead
df3['target_72h'] = df3['PM2.5'].shift(-72)   # PM2.5, 72 hours ahead

# printing
df3[['time', 'PM2.5', 'target_24h', 'target_48h', 'target_72h']].head(30)
```

```python
# Remove rows where we don't have a future value to predict
df_model = df3.dropna(subset=['target_24h', 'target_48h', 'target_72h']).reset_index(drop=True)

print("Before:", df.shape)
print("After dropping rows without targets:", df_model.shape)
```

```python
# Use 80% of the earliest data to train, last 20% to test
split_index = int(len(df_model) * 0.8)

train_df = df_model.iloc[:split_index]
test_df = df_model.iloc[split_index:]

print("Train size:", train_df.shape)
print("Test size:", test_df.shape)
```

```python
# Starting with just the 24h target 
feature_cols = ['PM10', 'hour', 'day_of_week', 'month', 
                'PM2.5_lag_1', 'PM2.5_lag_24', 
                'PM2.5_rolling_3', 'PM2.5_rolling_24']

X_train = train_df[feature_cols]
y_train = train_df['target_24h']

X_test = test_df[feature_cols]
y_test = test_df['target_24h']
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor

# Model 1: Linear Regression (draws a straight-line relationship)
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

# Model 2: Random Forest (learns complex patterns)
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

print("I trained both modelsss yayyyyyy!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

def evaluate_model(model, X_test, y_test, name):
    predictions = model.predict(X_test)
    mae = mean_absolute_error(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    r2 = r2_score(y_test, predictions)
    print(f"--- {name} ---")
    print(f"MAE:  {mae:.2f}  (average error in PM2.5 units)")
    print(f"RMSE: {rmse:.2f}  (penalizes big mistakes more)")
    print(f"R²:   {r2:.2f}  (1.0 = perfect, 0 = no better than guessing average)")
    print()

evaluate_model(lr_model, X_test, y_test, "Linear Regression")
evaluate_model(rf_model, X_test, y_test, "Random Forest")
```

```python
import joblib

# Save the best model (whichever had lower MAE/RMSE)
joblib.dump(rf_model, '../models/rf_model_24h.pkl')
print("Model saved!")
```

```python
#repeat traing for 48h and 72h target

y_train_48 = train_df['target_48h']
y_test_48 = test_df['target_48h']

y_train_72 = train_df['target_72h']
y_test_72 = test_df['target_72h']

rf_model_48 = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model_48.fit(X_train, y_train_48)

rf_model_72 = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model_72.fit(X_train, y_train_72)
```

```python
joblib.dump(rf_model, '../models/rf_model_24h.pkl')
joblib.dump(rf_model_48, '../models/rf_model_48h.pkl')
joblib.dump(rf_model_72, '../models/rf_model_72h.pkl')
```

```python
#Comparison Table
def evaluate_model(model, X_test, y_test, name):
    predictions = model.predict(X_test)
    mae = mean_absolute_error(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    r2 = r2_score(y_test, predictions)
    return {'Horizon': name, 'MAE': round(mae, 2), 'RMSE': round(rmse, 2), 'R2': round(r2, 2)}
```

```python
results = []
results.append(evaluate_model(rf_model, X_test, y_test, '24h'))
results.append(evaluate_model(rf_model_48, X_test, y_test_48, '48h'))
results.append(evaluate_model(rf_model_72, X_test, y_test_72, '72h'))

comparison_df = pd.DataFrame(results)
print(comparison_df)
```

```python
# Hyperparameter Check
# Try a few different settings on the 24h model only
settings_to_try = [
    {'n_estimators': 100, 'max_depth': None},
    {'n_estimators': 200, 'max_depth': 10},
    {'n_estimators': 200, 'max_depth': 20},
]

for settings in settings_to_try:
    model = RandomForestRegressor(**settings, random_state=42)
    model.fit(X_train, y_train)
    result = evaluate_model(model, X_test, y_test, str(settings))
    print(result)
```
